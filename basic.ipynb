```python
def train_model(model_name,*datasets,**hyperparameters):
    print(f"training model: {model_name}")
    print("\n Dataset")
    for dataset in datasets:
        print("-",dataset)
    print("\n hyperparameters")
    for key ,value in hyperparameters.item():
        print (f"{key}: {value}")
```

```python
train_model(
    "fraus detection",
    "train.csv",
    "validation.csv",
    "test.csv",
    learning_rate=0.001,
    epochs=100,
    batch_size=64,
    optimizer="Adam",
    early_stopping=True
)
```
